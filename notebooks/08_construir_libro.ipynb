```python
import os, shutil, zipfile, glob, subprocess
try:
    from google.colab import drive
    drive.mount('/content/drive')
except ImportError:
    pass

RUTA = '/content/drive/MyDrive/ML_LendingClub'
LIBRO = '/content/libro'
shutil.rmtree(LIBRO, ignore_errors=True)
with zipfile.ZipFile(RUTA + '/libro.zip') as z:
    z.extractall('/content')
assert os.path.isdir(LIBRO), 'libro.zip debe contener la carpeta libro/'

def copiar(origen, destino):
    os.makedirs(destino, exist_ok=True)
    n = 0
    for f in glob.glob(origen):
        if os.path.isfile(f):
            shutil.copy(f, destino); n += 1
    print(f'{n:2d} archivos  <-  {origen}')

R = LIBRO + '/resultados'
copiar(RUTA + '/resultados_sklearn.csv', R)
copiar(RUTA + '/resultados_pyspark.csv', R)
copiar(RUTA + '/resultados_eda/*', R + '/eda')
copiar(RUTA + '/resultados_pruebas/*', R + '/pruebas')
copiar(RUTA + '/resultados_lime/*', R + '/lime')
```

```python
r = subprocess.run(['jupyter-book', 'build', LIBRO], capture_output=True, text=True)
print(r.stdout[-1500:]); print(r.stderr[-2500:])
```

```python
html = LIBRO + '/_build/html'
assert os.path.exists(html + '/index.html'), 'No se generó el libro; revisa los mensajes anteriores'
salida = RUTA + '/libro_html'
shutil.make_archive(salida, 'zip', html)
print('Libro listo:', salida + '.zip')
```
